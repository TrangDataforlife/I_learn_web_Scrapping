# 📚 Tổng hợp HTTPBin Endpoints

## 1. HTTP Methods
| Endpoint | Chức năng | Ví dụ |
|---|---|---|
| `/get` | Trả về dữ liệu của GET request | `requests.get("https://httpbin.org/get")` |
| `/post` | Trả về dữ liệu của POST request | `requests.post("https://httpbin.org/post", json={"name": "trang"})` |
| `/put` | Trả về dữ liệu của PUT request | `requests.put("https://httpbin.org/put", json={"id": 1})` |
| `/patch` | Trả về dữ liệu của PATCH request | `requests.patch("https://httpbin.org/patch", json={"status": "done"})` |
| `/delete` | Trả về dữ liệu của DELETE request | `requests.delete("https://httpbin.org/delete")` |

## 2. Auth
| Endpoint | Chức năng | Ví dụ |
|---|---|---|
| `/basic-auth/:user/:passwd` | Yêu cầu xác thực HTTP Basic Auth | `requests.get("https://httpbin.org/basic-auth/user/pass123", auth=("user", "pass123"))` |
| `/hidden-basic-auth/:user/:passwd` | Basic Auth nhưng trả 404 nếu sai (ẩn) | `requests.get("https://httpbin.org/hidden-basic-auth/user/pass123", auth=("user", "pass123"))` |
| `/digest-auth/:qop/:user/:passwd` | Yêu cầu xác thực HTTP Digest Auth | `requests.get("https://httpbin.org/digest-auth/auth/user/pass123", auth=requests.auth.HTTPDigestAuth("user", "pass123"))` |
| `/bearer` | Kiểm tra Bearer token trong header, trả 401 nếu thiếu | `requests.get("https://httpbin.org/bearer", headers={"Authorization": "Bearer abc123"})` |

## 3. Status codes
| Endpoint | Chức năng | Ví dụ |
|---|---|---|
| `/status/:code` | Trả về mã trạng thái HTTP tùy chọn | `requests.get("https://httpbin.org/status/418")` |

## 4. Request inspection
| Endpoint | Chức năng | Ví dụ |
|---|---|---|
| `/headers` | Trả về toàn bộ headers gửi lên | `requests.get("https://httpbin.org/headers").json()` |
| `/ip` | Trả về IP nguồn (origin) của request | `requests.get("https://httpbin.org/ip").json()` |
| `/user-agent` | Trả về User-Agent của client | `requests.get("https://httpbin.org/user-agent").json()` |

## 5. Response inspection
| Endpoint | Chức năng | Ví dụ |
|---|---|---|
| `/response-headers?key=val` | Trả về response với headers tùy chỉnh | `requests.get("https://httpbin.org/response-headers?Content-Type=text/plain")` |
| `/cache` | Trả 200, hoặc 304 nếu có header `If-Modified-Since`/`If-None-Match` | `requests.get("https://httpbin.org/cache", headers={"If-None-Match": "abc"})` |
| `/cache/:n` | Set header `Cache-Control` trong n giây | `requests.get("https://httpbin.org/cache/60")` |
| `/etag/:etag` | Kiểm tra cơ chế ETag | `requests.get("https://httpbin.org/etag/abc123", headers={"If-None-Match": "abc123"})` |

## 6. Response formats
| Endpoint | Chức năng | Ví dụ |
|---|---|---|
| `/gzip` | Trả dữ liệu nén gzip | `requests.get("https://httpbin.org/gzip").json()` |
| `/deflate` | Trả dữ liệu nén deflate | `requests.get("https://httpbin.org/deflate").json()` |
| `/brotli` | Trả dữ liệu nén brotli | `requests.get("https://httpbin.org/brotli").json()` |
| `/html` | Trả về một trang HTML mẫu | `requests.get("https://httpbin.org/html").text` |
| `/xml` | Trả về dữ liệu dạng XML | `requests.get("https://httpbin.org/xml").text` |
| `/json` | Trả về dữ liệu JSON mẫu | `requests.get("https://httpbin.org/json").json()` |
| `/encoding/utf8` | Trả nội dung UTF-8 | `requests.get("https://httpbin.org/encoding/utf8").text` |

## 7. Dynamic data
| Endpoint | Chức năng | Ví dụ |
|---|---|---|
| `/uuid` | Sinh UUID4 ngẫu nhiên | `requests.get("https://httpbin.org/uuid").json()` |
| `/bytes/:n` | Sinh n byte dữ liệu ngẫu nhiên (nhận tham số `seed`) | `requests.get("https://httpbin.org/bytes/1024?seed=42").content` |
| `/stream-bytes/:n` | Stream n byte dữ liệu ngẫu nhiên | `requests.get("https://httpbin.org/stream-bytes/1024", stream=True)` |
| `/delay/:n` | Trì hoãn phản hồi n giây (test timeout) | `requests.get("https://httpbin.org/delay/3", timeout=5)` |
| `/drip?numbytes=n&duration=s&delay=s` | Nhỏ giọt dữ liệu theo thời gian (giả lập server chậm) | `requests.get("https://httpbin.org/drip?numbytes=5&duration=5&delay=1", stream=True)` |
| `/range/:n` | Trả về stream hỗ trợ byte-range request | `requests.get("https://httpbin.org/range/1024", headers={"Range": "bytes=0-100"})` |

## 8. Cookies
| Endpoint | Chức năng | Ví dụ |
|---|---|---|
| `/cookies` | Trả về cookie hiện có | `requests.get("https://httpbin.org/cookies", cookies={"token": "abc123"}).json()` |
| `/cookies/set?name=value` | Set một hoặc nhiều cookie | `s = requests.Session(); s.get("https://httpbin.org/cookies/set?token=abc123")` |
| `/cookies/delete?name` | Xóa một hoặc nhiều cookie | `s.get("https://httpbin.org/cookies/delete?token")` |

## 9. Images
| Endpoint | Chức năng | Ví dụ |
|---|---|---|
| `/image` | Trả về ảnh (tùy `Accept` header chọn định dạng) | `requests.get("https://httpbin.org/image", headers={"Accept": "image/png"}).content` |
| `/image/png`, `/image/jpeg`, `/image/webp`, `/image/svg` | Trả về ảnh theo định dạng cụ thể | `requests.get("https://httpbin.org/image/png").content` |

## 10. Redirects
| Endpoint | Chức năng | Ví dụ |
|---|---|---|
| `/redirect/:n` | Redirect 302 n lần liên tiếp | `requests.get("https://httpbin.org/redirect/3")` |
| `/redirect-to?url=...` | Redirect 302 đến URL chỉ định | `requests.get("https://httpbin.org/redirect-to?url=https://example.com")` |
| `/relative-redirect/:n` | Redirect tương đối n lần | `requests.get("https://httpbin.org/relative-redirect/3")` |
| `/absolute-redirect/:n` | Redirect tuyệt đối n lần | `requests.get("https://httpbin.org/absolute-redirect/3")` |

## 11. Anything
| Endpoint | Chức năng | Ví dụ |
|---|---|---|
| `/anything` và `/anything/:anything` | Trả về nguyên vẹn mọi dữ liệu (method, headers, body, args...) được gửi lên | `requests.post("https://httpbin.org/anything", json={"hello": "world"}).json()` |

## 12. Other Utilities
| Endpoint | Chức năng | Ví dụ |
|---|---|---|
| `/robots.txt` | Trả về nội dung robots.txt mẫu | `requests.get("https://httpbin.org/robots.txt").text` |
| `/deny` | Trang bị chặn bởi robots.txt (test respect robots) | `requests.get("https://httpbin.org/deny").text` |
| `/links/:n` | Trả về trang chứa n link HTML | `requests.get("https://httpbin.org/links/10").text` |
| `/forms/post` | Trả về HTML form để submit tới `/post` | `requests.get("https://httpbin.org/forms/post").text` |
| `/base64/:value` | Decode chuỗi Base64 | `requests.get("https://httpbin.org/base64/SFRUUEJJTiBpcyBhd2Vzb21l").text` |
